```python
from pathlib import Path

import pandas as pd

p_ehr = list(Path('/workspaces/BrainFlux/phantom_menace/EHR').rglob('*.csv'))
p_meds = list(Path('/workspaces/BrainFlux/phantom_menace/med').rglob('*.csv'))

p_meds.sort()
p_ehr.sort()

print(f'Found {len(p_ehr)} EHR files and {len(p_meds)} medication files.')

get_id = lambda p: p.stem.split('_')[0]



df_merged_list = []

for file_ehr, file_med in zip(p_ehr, p_meds):
    assert get_id(file_ehr) == get_id(file_med), f"ID mismatch: {file_ehr} and {file_med}"
    
    df_ehr = pd.read_csv(file_ehr)
    df_med = pd.read_csv(file_med)
    
    df_ehr.drop(columns=['normalized_id'], inplace=True, errors='ignore')
    df_ehr.rename(columns={'result': 'result/dose'}, inplace=True)
    df_ehr['route'] = 'N/A'
    df_ehr.rename(columns={'name': 'name/drug'}, inplace=True)
    
    df_med.rename(columns={'dose': 'result/dose'}, inplace=True)
    df_med.rename(columns={'normalized_id': 'id'}, inplace=True)
    df_med.rename(columns={'drug': 'name/drug'}, inplace=True)
    
    # print('EHR columns:', df_ehr.columns.tolist())
    # print('Med columns:', df_med.columns.tolist())
    
    
    df_merged = pd.concat([df_ehr, df_med], ignore_index=True)
    df_merged.sort_values(by='relative_sec', inplace=True)
    
    
    print(f'ID {get_id(file_ehr)} has {len(df_merged)} records.')
    
    # df_merged.to_csv(f'/workspaces/BrainFlux/phantom_menace/{get_id(file_ehr)}_merged.csv', index=False)
    df_merged_list.append(df_merged)
    
df_all = pd.concat(df_merged_list, ignore_index=True)
df_all.to_csv('/workspaces/BrainFlux/phantom_menace/all_merged.csv', index=False)
print(f'Saved all merged data with {len(df_all)} total records.')
    
```
